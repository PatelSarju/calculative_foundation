- Load the dataset

```python
import pandas as pd
import numpy as np
import scipy as sp
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
```

```python
score_data = pd.read_csv("student_scores.csv")
```

* Tasks are given below

- Part A: Vector & Matrix Fundamentals
    1. Represent each student's subject scores as a vector.
    2. Compute:
        - Norm-1 and Norm-2 of vectors.
        - Dot product and angle between two students' score vectors.
        - Cross product (for 3D selected subjects).
    3. Find the projection of one vector onto another.

```python
# Represent each student's subject scores as a vector
student_score = score_data.to_numpy()
```

```python
# Compute norm-1 and norm-2 of vectors

# find the norm-1 (manhattan distance)
np.linalg.norm(student_score, ord=1, axis=1)

# find the norm-2 (euclidean distance)
np.linalg.norm(student_score, ord=2, axis=1)
```

```python
# Dot product and angle between two students' score vectors.

# pick any two students
student_1 = student_score[0]
student_2 = student_score[1]

# perform the dot product
dot_product = np.dot(student_1, student_2)

# multiply the distance of both vectors
student_1_distance = np.linalg.norm(student_1, ord=2)
student_2_distance = np.linalg.norm(student_2, ord=2)
multiplied_distance = student_1_distance * student_2_distance

# find the angle
ang_radians = np.arccos(dot_product / multiplied_distance)

# convert radians to degree
ang_degree = np.degrees(ang_radians)

print(f"Dot product of two students score is {dot_product}")
print(f"Angle between two students score is {ang_degree}")
```

```python
# Cross product (for 3D selected subjects).

# pick any two students
student_1 = student_score[0,:3]
student_2 = student_score[1,:3]

# perform the cross product
cross_product = np.cross(student_1, student_2)

print(f"Cross Product of two student's three subject is {cross_product}")
```

```python
# Find the projection of one vector onto another.

# pick any two students
student_1 = student_score[0]
student_2 = student_score[1]

# find the dot product
dot_product = np.dot(student_1, student_2)

# find the euclidean distance of student_2
distance = np.linalg.norm(student_2, ord=2)
distance_sqr = distance**2

# find the projection of student 1 on student 2
projection_of_1_on_2 = (dot_product / distance_sqr) * student_2

print(f"Projection of student 1 on student 2 is {projection_of_1_on_2}")
```

- Part D: Eigenvalues & Decomposition
    - Compute the eigenvalues and eigenvectors of the covariance matrix.
    - Perform LU Decomposition of the dataset matrix.
    - Perform Singular Value Decomposition (SVD) and explain its role in dimensionality
reduction.

```python
# Compute the eigenvalues and eigenvectors of the covariance matrix.

# form a covariance matrix
covariance_matrix = np.cov(student_score, rowvar=False)

# find the eigen_values and eigen_vectors
eigen_values, eigen_vectors = np.linalg.eig(covariance_matrix)

print("Covariance matrix are given below")
for i in covariance_matrix:
    print(i)
    
print("\n\nEigen Values are given below")
for i in eigen_values:
    print(i)

print("\n\nEigen Vectors are given below")
for i in eigen_vectors:
    print(i)
```

```python
# Perform LU Decomposition of the dataset matrix.

# derive the squeare matrix from the given dataset matrix
score_square_matrix = student_score[:6,:]

# perform LU decomposition on square matrix
permutation_matrix, lower_triangular_matrix, upper_triangular_matrix = sp.linalg.lu(score_square_matrix)

print(f"Permutation matrix is given below")
for i in permutation_matrix:
    for j in i:
        print(j, end="  ")
    print()
print("\n\nLower Triangular matrix is given below")
for i in lower_triangular_matrix:
    for j in i:
        print(round(j,2), end="  ")
    print()
    
print(f"\n\nUpper Triangular matrix is given below")
for i in upper_triangular_matrix:
    for j in i:
        print(round(j,2), end="  ")
    print()
```

```python
# Perform Singular Value Decomposition (SVD) and explain its role in dimensionality reduction.

# perform SVD
U, S, V_t = np.linalg.svd(student_score, full_matrices=False)

print(f"U matrix (left singular vectors) shape: {U.shape}")
print(f"S (singular values): {S}")
print(f"V_t matrix (right singular vectors) shape: {V_t.shape}")


# Role of SVD in Dimensionality Reduction:

# SVD breaks data into parts. By keeping only the most important parts (components), we reduce the number of features while keeping most of the information. This makes the data simpler and easier to work with.
```

- Part E: Dimensionality Reduction
    - Apply Principal Component Analysis (PCA) to reduce the dataset from multiple
    subjects to 2 dimensions.
    - Apply Linear Discriminant Analysis (LDA) to classify students into "Above Average"
    and "Below Average" categories.

```python
# Apply Principal Component Analysis (PCA) to reduce the dataset from multiple subjects to 2 dimensions.

# standardize the data
scaler = StandardScaler()
standard_data = scaler.fit_transform(student_score)

# apply PCA to reduce 2 dimensions
pca = PCA(n_components = 2)
standard_pca = pca.fit_transform(standard_data)

# Display the first two principal components for each student
print("PCA - 2D projection of the data:")
print(standard_pca[:, :])
```

```python
# Apply Linear Discriminant Analysis (LDA) to classify students into "Above Average" and "Below Average" categories.

# Create a "class" label for each student based on average score
average_score = np.mean(student_score, axis=1)
threshold = np.median(average_score)  # Set threshold as the median score

# Classify as Above Average (1) or Below Average (0)
y = np.where(average_score > threshold, 1, 0)

# Apply LDA
lda = LinearDiscriminantAnalysis(n_components=1)
X_lda = lda.fit_transform(standard_data, y)

# Display the LDA transformed data 
print("LDA - Student classification into Above Average/Below Average (first 5 students):")
print(X_lda[:, :])

# Display the class labels
print("\nClass labels (Above Average: 1, Below Average: 0):")
print(y[:])
```
